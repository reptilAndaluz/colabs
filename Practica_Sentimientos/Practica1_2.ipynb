```python
from pysentimiento import create_analyzer
import re

LISTA_NEGRA = ["horrible", "odio", "asco", "estafa", "pésimo", "fatal", "desastre", "peor"] # edita aquí

analizador = create_analyzer(task="sentiment", lang="es")  # RoBERTuito




def es_valido(mensaje):
    palabras = re.findall(r"\w+", mensaje.lower())
    if any(p in LISTA_NEGRA for p in palabras):
        return False
    return analizador.predict(mensaje).output != "NEG"


mensajes = [
    # Positivas
    "Me encantó el servicio, volveré sin duda",
    "Qué día tan bueno, estoy muy feliz",
    "La comida estaba deliciosa",
    "Gracias por la ayuda, fuiste muy amable",
    "Es el mejor producto que he comprado",
    "Todo salió perfecto, estoy muy contento",
    "Qué bonito lugar, me ha encantado",
    "Excelente atención, muy profesionales",
    "Me siento genial después del concierto",
    "Lo recomiendo al cien por cien",
    # Negativas
    "Esto es horrible, lo odio",
    "El peor servicio que he recibido en mi vida",
    "Qué asco de comida, no vuelvo más",
    "Me han tratado fatal, estoy muy enfadado",
    "Es una estafa, no compréis esto",
    "Llegó roto y nadie me ayuda, qué desastre",
    "Qué día tan malo, todo me sale mal",
    "Pérdida de tiempo y de dinero, muy decepcionante",
    "Odio esta aplicación, no funciona nunca",
    "Un trato pésimo, nunca más",
]

for m in mensajes:
    estado = " válido  " if es_valido(m) else "excluido"
    print(f"{estado} | {m}")
validos = [m for m in mensajes if es_valido(m)]
print(*validos)
```
