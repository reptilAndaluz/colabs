# **K-means**

**K-means (k-medias)** es un algoritmo de aprendizaje automático no supervisado que divide un conjunto de datos en k grupos o clústeres basándose en la similitud de sus características.

Su principal limitación es que el usuario debe definir el valor de k inicialmente.

Este algoritmo se usa cuando tenemos datos sin etiquetas. Por ejemplo, medidas de flores (largo de pétalo, ancho de pétalo) pero no se sabe qué tipos de flores son. Se quiere descubrir cuántos tipos de flores hay.

En realidad, sí se conocen las tres especies, que luego usaremos para comparar el resultado obtenido: Iris **setosa**, Iris **versicolor** e Iris **virginica**.

```python
from sklearn.cluster import KMeans

# === este es el número de grupos ===
n_clusters = 3
# =====================================

# se configura el modelo
modelo = KMeans(n_clusters=n_clusters, random_state=0, n_init=10)
# se entrena el modelo
modelo.fit(datos)
```

```python
print(modelo.labels_)          # etiqueta de cluster (0, 1, 2...) de cada muestra
print(modelo.cluster_centers_) # coordenadas de los K centroides (shape: K × n_features)
print(modelo.inertia_)         # suma de distancias² (cómo de "compactos" son los clusters)
print(modelo.n_iter_)          # iteraciones que necesitó para converger
```

```python
import matplotlib.pyplot as plt

plt.scatter(datos[:, 0], datos[:, 1], c=modelo.labels_, cmap="viridis")
plt.title(f"Agrupación con {n_clusters} grupos")
plt.show()
```

```python
# aquí va la implementación

import numpy as np
from sklearn.metrics import confusion_matrix, accuracy_score

# 1. Ingesta de Vectores de Datos
datos_reales = iris.target
datos_predichos = modelo.labels_

# 2. Lógica de Alineación de Etiquetas (Workaround para K-Means)
etiquetas_alineadas = np.zeros_like(datos_predichos)

for i in range(3):
    # Aislar el subgrupo asignado por K-Means
    mascara = (datos_predichos == i)
    if np.any(mascara):
        # Mapear el ID del clúster a la etiqueta real predominante
        etiqueta_real_asociada = np.bincount(datos_reales[mascara]).argmax()
        etiquetas_alineadas[mascara] = etiqueta_real_asociada

# 3. Auditoría de Datos y Porcentajes
porcentaje_acierto = accuracy_score(datos_reales, etiquetas_alineadas) * 100
matriz_fallos = confusion_matrix(datos_reales, etiquetas_alineadas)

# 4. Consola de Rendimiento
print("=== REPORTE DE TRANSFORMACIÓN ===")
print(f"Precisión Global de la IA: {porcentaje_acierto:.2f}%\n")

print("Matriz de Confusión (Desglose Estructural):")
print("  Filas: Datos de Verdad | Columnas: Datos Obtenidos")
print(matriz_fallos, "\n")

print("--- Desglose de Fallos por Categoría ---")
for i in range(3):
    total = np.sum(datos_reales == i)
    aciertos = matriz_fallos[i, i]
    fallos = total - aciertos
    porcentaje_error_cat = (fallos / total) * 100 if total > 0 else 0
    print(f"Categoría Original {i}: {fallos} fallos registrados ({porcentaje_error_cat:.1f}% de error sobre el total del grupo).")

    """
=== SCRIPT DE EVALUACIÓN ===

Este script realiza la comparación entre los datos de verdad y los datos generados por el modelo de IA.
Su objetivo es mitigar el desplazamiento de índices natural de K-Means y arrojar estadísticas de precisión.

Ingesta de Vectores de Datos:
   - Se importan los datos originales (iris.target) y las etiquetas generadas por el modelo (modelo.labels_).

Lógica de Alineación de Etiquetas (Workaround):
   - K-Means no asigna los mismos IDs a los grupos que el conjunto de datos original (ej: el grupo 0 predicho puede ser el grupo 1 real).
   - Se itera sobre los clústeres generados (0, 1, 2) y se usa np.bincount para encontrar cuál es la etiqueta real más frecuente (argmax) dentro de cada grupo predicho.
   - Con esta lógica, se "remapean" o alinean las predicciones para que tengan el mismo identificador que los datos reales.

Auditoría de Datos y Porcentajes:
   - Se calcula el porcentaje global de acierto usando `accuracy_score`.
   - Se genera una matriz de confusión con `confusion_matrix`, la cual cruza las etiquetas reales (filas) con las predichas (columnas) para identificar cruces anómalos.

Consola de Rendimiento:
   - Se imprime la precisión global.
   - Se imprime la matriz para un análisis visual.
   - Se itera sobre cada categoría (0, 1, 2) para desglosar cuántos errores específicos cometió el modelo dentro de esa clase y qué porcentaje representan esos errores frente al total de muestras de la categoría.
"""
```
